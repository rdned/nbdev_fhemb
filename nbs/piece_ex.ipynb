### Specify the `Piece` object, which stores:

- the opera title,
- the time interval (as a frame range),
- subject intervals, where each subject is associated with a tracked heatmap time series.

```python
#| export

dg0 = Piece(              
    title='Don Giovanni',
    time_interval=(14200,15200), 
    subjs=[(40, 42), 50, (65,68)]
) 
```

### Ask for an attribute
  - `facebbox_t` will be fetched from a DB 

```python
#| echo: true
#| eval: false

dg0.facebbox_t
```

### Now the attribute is cached

```python
#| eval: false

dg0.api()
```

>`facebbox_t` is a `Datasource` object (`fhemb.dbms.db.Datasource`) used to access tracked face-bounding-box time-series data and derived embeddings. In this example, it contains 7 tracked time series of subjects $\{40,41,42,50,65,67,68\}$ who enjoyed *Don Giovanni* opera. All the time series are restricted to the segment of $1000$ frames in the specified interval $[14200, 15200)$.
