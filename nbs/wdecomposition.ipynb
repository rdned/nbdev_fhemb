# `Wdecomposition` API
> `Wdecomposition` encapsulates the wavelet‑based decomposition of an `Embedding`, exposing each frequency‑band subsignal as its own `Embedding` and providing methods for their analysis and recombination.

#### The Method Resolution Order (MRO) 
>the exact linear path Python follows when looking up attributes or when super() is invoked.

```raw
Wdecomposition
 └── ValidateSubjsMixin
      └── ValidateFbandsMixin
           └── object
```

#### Architectural responsibilities

- **Frequency‑domain bridge**: decompose an `Embedding` into band‑limited components and map each band to a band‑specific `Embedding`.

- **Metadata and caching**: record metadata about the wavelet family and decomposition level, numeric frequency and period ranges, and cache reconstructions to avoid recomputation.

- **Loop closer**: produce new `Embedding` objects by reconstructing embeddings from selected bands (so the pipeline can iterate: `Embedding` → `Wdecomposition` → `Embedding` → ...). `Wdecomposition` is transient — it enables multiresolution analysis but the `Embedding` remains the canonical product the system analyzes.

````python
show_doc(Wdecomposition, name='The Constructor', title_level=2)
````

## The Core Method

**`create_wreconstruction(subjs, fbands)` → `Embedding`**

Reconstructs an embedding from a selected set of frequency bands.

- If `fbands` is empty → returns the original decomposed embedding
- Otherwise → returns a reconstruction using only the specified bands
- In both cases → projected to the requested subjects

````python
show_doc(Wdecomposition.create_wreconstruction, name='create_wreconstruction', title_level=3)
````
