# Create Complex Embeddings
> by constructing an `EmbeddingCollection` object

There are two routes to construct an `EmbeddingCollection` object. One is to use the `EmbeddingCollection` constructor, and the other is to use the `Datasource.projection` syntactic shorthand. We demonstrate both: first constructing *full embeddings*, i.e., without using projection, then constructing the more general *projected embeddings*.    

## Full Embeddings
> no features are projected out

```python
#| exports
#| eval: false

pca_emb = dg0.face_t.create_pca_embedding(   # embed into principal components
    factory=pca                              # specify PCA factory to use for creating the PCA embedding (imported)
)[0]                                         # the function 'create_pca_embedding' return a tuple (Embedding, ...)

percentile_emb = dg0.face_t.percentiles_ts   # embed into percentiles  

bin_emb = dg0.face_t.binheights_ts           # embed into bin heights
```

### Combine full embeddings 

#### `EmbeddingCollection` constructor

```python
#| exports
#| eval: false

combined_emb0 = EmbeddingCollection(pca_emb, percentile_emb, bin_emb)
```

#### `Datasource.project` syntactic shorthand

```python
#| exports
#| eval: false

combined_emb1 = dg0.face_t.project(                              # project all available features of the Datasource face_t to specified `features`
    features=[('pc0', 'pc4'), ('p0', 'p100'), ('bh0', 'bh18')],  # specify all features: none will be projected out
    pcfactory=pca,                                               # specify PCA for principle components 
)
```

#### `EmbeddingCollection` constructor and `Datasource.project` method create the same combined `EmbeddingCollection` object

```python
#| eval: false

combined_emb0.features
```

```python
#| eval: false
combined_emb1.features
```

```python
#| eval: false

combined_emb0.features_tint['subj40']['pcomponents_ts'] == combined_emb1.features_tint['subj40']['pcomponents_ts']
```

```python
#| eval: false

combined_emb0.get_ts_subjs('pc1')['pcomponents_ts'] - combined_emb1.get_ts_subjs('pc1')['pcomponents_ts']
```

## Projected Embeddings
> only selected features are included in the embedding

```python
#| exports
#| eval: false

pca_emb_prj = pca_emb.project(         # project all available features of the Embedding pca_emb to specified `features`    
    subjs=[40],                        # specify subject(s) to project to: may be specified also as intervals, e.g., [(40, 45), 47]   
    features=['pc0', 'pc1'],           # specify features to project to: principal components are prefixed by 'pc', may be specified also as intervals, e.g., [('pc0', 'pc4')] 
)

percentile_emb_prj = percentile_emb.project(   # project all available features of the Embedding percentile_emb to specified `features`  
    subjs=[40],  
    features=['p0', 'p100'],           # percentiles are prefixed by 'p', may be specified also as intervals, e.g., [('p0', 'p20'), ('p80', 'p100')] 
)

bin_emb_prj = bin_emb.project(         # project all available features of the Embedding bin_emb to specified `features`     
    subjs=[40], 
    features=[('bh0', 'bh10')],        # bin heights are prefixed by 'bh': bins 1,2,..,10
)
```

### Combine projected embeddings

#### `EmbeddingCollection` constructor

```python
#| exports
#| eval: false

combined_emb0_prj = EmbeddingCollection(pca_emb_prj, percentile_emb_prj, bin_emb_prj)
```

#### `Datasource.project` syntactic shorthand

```python
#| exports
#| eval: false

combined_emb1_prj = dg0.face_t.project(       # project all available features of the Datasource face_t to specified `features`  
    subjs=[40],                               # specify subject(s) to project to: may be specified also as intervals, e.g., [(40, 45), 47] 
    features=['pc0', 'pc1', 'p0', 'p100', ('bh0', 'bh10')],  # specify features to project to: principal components 'pc', percentiles 'p', bin heights 'bh'   
    pcfactory=pca,                            # specify PCA for principle components
)
```

#### `EmbeddingCollection` and `Datasource.project` create the same combined `Embedding`

```python
#| eval: false

combined_emb0_prj.features
```

```python
#| eval: false

combined_emb1_prj.features
```

```python
#| eval: false

combined_emb0_prj.features_tint['subj40']['percentiles_ts']
```

```python
#| eval: false

combined_emb0_prj.features_tint['subj40']['pcomponents_ts'] == combined_emb1_prj.features_tint['subj40']['pcomponents_ts']
```

```python
#| eval: false

combined_emb1_prj.get_ts_subjs('bh5')['binheights_ts'] - combined_emb1_prj.get_ts_subjs('bh5')['binheights_ts']
```
