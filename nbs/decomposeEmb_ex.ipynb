# Decompose Embedding

## Create a `Wdecomposition` object

```python
#| eval: false
#| exports

percentile_emb_decomp = percentile_emb.create_wdecomposition(  # decompose to frequency bands of the wavelet transform specified by 'factory'
    features=['p0', 'p50', 'p90'],     # specify features of interest: percentiles are prefixed by 'p' and may be specified also as intervals, e.g., [('p0, 'p20)]
    factory=wfactory('db4',8)          # specify the wavelet transform for the decomposition
)
```

## Inspect the `Wdecomposition` object

### Metadata

#### Choose a frequency band

```python
#| eval: false

frequency_band = 0   # specify the frequency band of the transform
```

##### Period in seconds

```python
#| eval: false

percentile_emb_decomp.periods[frequency_band]   # show the period interval of the frequency band
```

##### Fequency in Hz

```python
#| eval: false

percentile_emb_decomp.frequencies[frequency_band]  # show the frequency interval of the frequency band 
```

### Data

##### The `Wdecomposition.fband_embedding`
> returns the Embedding object: 
> $\mathrm{fbands} \rightarrow \mathrm{Embedding\ objects}$

```python
#| eval: false

percentile_emb_decomp.fband_embedding[frequency_band].features_tint['subj40'].shape  # show the signal shape of the specified subject and frequency band 
```

##### The `Wdecomposition.subsignals`
> returns a mapping directly into the data:
> $\mathrm{subjects} \times \mathrm{fbands} \rightarrow \mathrm{decomposed\ signals}$

```python
#| eval: false

percentile_emb_decomp.subsignals['subj40'][frequency_band].shape  # show the signal shape of the specified subject and frequency band 
```

Note that the above contain the same signal.

```python
#| eval: false 

percentile_emb_decomp.fband_embedding[frequency_band].features_tint['subj40'] == percentile_emb_decomp.subsignals['subj40'][frequency_band]
```
