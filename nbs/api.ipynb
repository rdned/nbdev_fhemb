# API
>Here we present the basic functionality of the fhemb

It's architecture implements a pipeline of four classe: `Piece` → `Datasource` → `Embedding` → `Decomposition`, where `Piece` and `Datasource` are input providers (the `Piece` supplies metadata, `Datasource` supplies raw content), `Embedding` implements canonical signal objects with provenance and analysis methods, and multi‑scale transforms yielding `WDecomposition`, which enables inverse reconstruction back to a denoised `Embedding`.


```raw
Piece (face_t, facebbox_t)
 └── Datasource (binheights_ts, percentiles_ts, features_ts)
      └── Embedding (feature × time per subject)
           └── Wdecomposition (bands → band-Embeddings)
                └── Embedding (reconstructed; loop back to Wdecomposition)
```
