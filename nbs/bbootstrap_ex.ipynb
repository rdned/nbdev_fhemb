# Block Bootstrap Embeddings
> Any `Embedding` object can be block bootstrapped into another `Embedding` object

::: {.callout-note}
Note that `EmbeddingCollection` invokes `.plot_signal` and `.bbootstrap` methods through its [Dynamic Attribute Forwarding](embeddingColl.ipynb#dynamic-attribute-access).
:::

```python
#| eval: false

from nbdev_fhemb.embedding_ex import pca_emb, combined_emb0
```

## The Bootstrapped Signal 

```python
#| eval: false

pca_emb.plot_signal(
    subjs=[40,41],  # specify subject(s) to plot: may be specified also as intervals, e.g., [(40, 45), 47]
    feature='pc2'   # specify feature to plot: principal components are prefixed by 'pc' 
)
```

```python
#| eval: false

combined_emb0.plot_signal(
    subjs=[40,41],  # specify subject(s) to plot: may be specified also as intervals, e.g., [(40, 45), 47]
    feature='pc2'   # specify feature to plot: principal components are prefixed by 'pc'
)
```

## Bootstrap with Different Block Lengths
>Here we demonstrate the bootstrap method, applied to an `Embedding` and to an `EmbeddingCollection` respectively. 

```python
#| eval: false

pca_emb.bbootstrap(
    block_length=5, # specify the length of the block in the block bootstrap
    rseed=0,        # random seed for reproducibility
).plot_signal(
    subjs=[40,41],  # specify subject(s) to plot: may be specified also as intervals, e.g., [(40, 45), 47]
    feature='pc2'   # specify feature to plot: principal components are prefixed by 'pc' 
)
```

```python
#| eval: false

combined_emb0.bbootstrap(
    block_length=5, # specify the length of the block in the block bootstrap
    rseed=0,        # random seed for reproducibility
).plot_signal(
    subjs=[40,41],  # specify subject(s) to plot: may be specified also as intervals, e.g., [(40, 45), 47]
    feature='pc2'   # specify feature to plot: principal components are prefixed by 'pc'
)
```

```python
#| eval: false

pca_emb.bbootstrap(
    block_length=10, # specify the length of the block in the block bootstrap
    rseed=0,         # random seed for reproducibility
).plot_signal(
    subjs=[40,41],   # specify subject(s) to plot: may be specified also as intervals, e.g., [(40, 45), 47]
    feature='pc2'    # specify feature to plot: principal components are prefixed by 'pc' 
)
```
