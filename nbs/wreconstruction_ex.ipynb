## Explicit Construction via `Wdecomposition` and `EmbeddingCollection` Constructor
> exposes the wavelet pipeline step by step: first creating a decomposition embedding, then deriving a reconstruction embedding from selected subbands. This form makes each analytic stage visible and individually inspectable.

### `Piece.Datasource.Embedding.Wdecomposition.create_reconstruction`

::: {.callout-note}
`Wdecompostion.create_reconstrction` returns an `Embedding` object
:::

```python
#| eval: false
#| exports

percentile_emb_prj_decomp_rec = dg0.face_t.percentiles_ts.create_wdecomposition(  # decompose to frequency bands of the wavelet transform specified by 'factory'
    features=['p0', 'p100'],                                                  # specify features of interest: percentiles are prefixed by 'p' and may be specified also as intervals, e.g., [('p0, 'p20)]
    factory=wfactory('db4',8)                                                 # specify the wavelet transform for the decomposition
).create_wreconstruction(                                                     # reconstruct a signal from specified frequency bands (fbands)
    subjs=[(40, 42), 50],                                                     # Subjects to project the reconstruction for: may be specified as intervals, e.g., [(40, 42)]
    fbands=[2,3,4]                                                            # frequency bands to include in the reconstruction (default is 'all')
)
```

```python
#| eval: false
#| exports

bin_emb_prj_decomp_rec = dg0.face_t.binheights_ts.create_wdecomposition(  # decompose to frequency bands of the wavelet transform specified by 'factory'
    features=[('bh0', 'bh10')],                                       # specify features of interest: bin heights are prefixed by 'bh and may be specified also as intervals, e.g., [('bh0, 'bh10)]
    factory=wfactory('db4',8)                                         # specify the wavelet transform for the decomposition
).create_wreconstruction(                                             # reconstruct a signal from specified frequency bands (fbands)
    subjs=[(40, 42), 50],                                             # Subjects to project the reconstruction for: may be specified as intervals, e.g., [(40, 42)]    
    fbands=[2,3,4]                                                    # frequency bands to include in the reconstruction (default is 'all')
)   
```

##### Characteristics

- Transparent pipeline: decomposition and reconstruction appear as distinct, composable transformations.

- Embeddings as first‑class objects: each step produces a concrete Embedding with its own provenance.

- Maximal flexibility: intermediate embeddings can be inspected, reused, or branched.

::: {.callout-tip}
This path is ideal when documenting or debugging the analytic flow.
:::

## Concise Construction via `Datasource.project`
>is a syntactic shorthand that collapses decomposition, reconstruction, and optional PCA into a single declarative call. It produces the same final embedding without exposing intermediate stages.

Construct the same wavelet‑based subband embedding as in the explicit `create_wdecomposition` → `create_wreconstruction` pipeline, but using the
compact `project(...)` shorthand.  
    
This single call internally:

1. applies PCA (if `pcfactory` is provided),

2. applies wavelet decomposition using `wdfactory`,

3. reconstructs from the selected frequency bands (`fbands`),

4. returns the final embedding for the requested subjects.

In contrast to the explicit step‑by‑step construction, no intermediate embeddings are materialized; the entire pipeline is specified declaratively.

```python
#| eval: false
#| exports

emb_prj_decomp_rec1 = dg0.face_t.project( 
    subjs=[(40, 42), 50],                 # Subjects to project the reconstruction for;
                                          # intervals may be used, e.g. [(40, 42)].

    pcfactory=pca,                        # Optional PCA stage: if provided, principal
                                          # components become available as features
                                          # (pc0, pc1, …). If omitted, no PCA is applied.

    wdfactory=wfactory('db4', 8),         # Wavelet transform factory used for the
                                          # decomposition stage; identical to the one
                                          # used in the explicit examples.

    fbands=[2, 3, 4],                     # Frequency bands to include in the reconstruction
                                          # (default is 'all'); same semantics as in
                                          # create_wreconstruction(...).

    features=['pc0', 'pc1', 'p0', 'p100', ('bh0', 'bh10')],
                                          # All features to retain in the embedding.
                                          # Any feature omitted here is projected out.
                                          #
                                          # Feature prefixes:
                                          #   - 'pc'  → principal components
                                          #   - 'p'   → percentiles
                                          #   - 'bh'  → bin heights
                                          #
                                          # Intervals may be specified as tuples,
                                          # e.g. ('bh0', 'bh10').
)
```

##### Characteristics

- Single entry point: the entire wavelet‑projection pipeline is specified at once.

- No intermediate artifacts: decomposition and reconstruction occur internally.

- User‑facing convenience: ideal for routine workflows where the pipeline is known and stable.

::: {.callout-tip}
This path favors readability and compactness over explicit staging.
:::
