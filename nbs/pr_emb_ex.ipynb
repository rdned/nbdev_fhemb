# Pairwise Relation Embedding

> Computed over embedded time series (*primary embeddings*) as a vector of pairwise relations between a given embedded time series and all other embedded time series.

---



### What it is

**Pairwise relation embedding** is a **secondary embedding layer** that encodes how one embedded time series relates to all others.  
It operates on *primary embeddings* - fixed‑dimensional vectors representing temporal patterns such as trend, variability, and autocorrelation - and computes a vector of similarity or distance values.

This produces a new representation:

- **Input:** primary embedding of series $i$  
- **Output:** vector 

  $[\, r(i,1),\ r(i,2),\ \ldots,\ r(i,N) \,]$  

  where each $r(i,j)$ quantifies the relation between series $i$ and series $j$.

Typical relation functions include **cosine similarity**, **Euclidean distance**, and **correlation**.

---

### Why it exists

Primary embeddings describe *what* each time series looks like.  
Pairwise relation embeddings describe **how each series compares to all others**, enabling downstream tasks that depend on global structure rather than isolated features.

They are useful when:

- **Global context matters** (e.g., clustering, grouping sensors by behavior)
- **Similarity‑based reasoning** is required (e.g., nearest‑neighbor classification)
- **Graph‑based models** need edge weights derived from embedding similarity

This layer exposes the relational geometry of the dataset, which is often more informative than per‑series features alone.

---

### How it is computed

Although terminology varies across the literature, the computational pattern is consistent:

1. **Compute primary embeddings**  using [FhEmb API](./api.ipynb)
   Each time series is mapped to a fixed‑dimensional vector capturing its temporal characteristics.

2. **Compute pairwise relations**  
   For a target embedding $i$, compute relations to all others $j$, $j = 1,\ldots, N$, using a chosen metric.

3. **Assemble the relation vector**  
   The resulting vector has length $N$ and can be used directly or passed to higher‑level models such as  
   **graph neural networks**, **clustering algorithms**, or **meta‑embedding layers**.

---

### Intuition

If primary embeddings encode **intrinsic properties**, pairwise relation embeddings encode **relational structure**.

Analogies:

- Word embeddings -> similarity matrix  
- Image embeddings -> neighborhood graph  
- Sensor networks -> correlation graph  

This layer answers: *“How does this series sit inside the global embedding space?”*

---

### When to use it

Pairwise relation embeddings are especially effective when:

- You need **dataset‑level structure**, not just per‑series features  
- You require **robust similarity signals** for classification or anomaly detection  
- You are constructing **graph‑based models** where edges represent similarity
