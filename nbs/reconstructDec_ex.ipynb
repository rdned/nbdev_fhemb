# Reconstruct Embedding from a Wavelet Band-Interval 
> Reconstruct a signal from specified frequency bands (`fbands`)

::: {.callout-note}
`fbands='all'` is equivalent to `fbands=[0,1,2,3,4,5,6,7,8]` when the `Wdecomposition` onbject has been created with a wavelet transform of level 8.
Using `fbands='all'` reconstructs the full signal from all subbands.  
:::

```python
#| eval: false

from nbdev_fhemb.decomposeEmb_ex import percentile_emb_decomp
```

```python
#| eval: false
percentile_emb_decomp.create_wreconstruction( # reconstruct a signal from specified frequency bands (fbands)
    subjs=[(40, 42), 50],                     # Subjects to project the reconstruction for: may be specified as intervals, e.g., [(40, 42)]
    fbands='all'                              # frequency bands to include in the reconstruction (default is 'all')
).features_tint['subj40']
```

::: {.callout-note}
Using `fbands = []` disables wavelet-based reconstruction and returns the original (undecomposed) input signal unchanged.
:::

::: {.callout-warning}
Returning the original (undecomposed) input signal bypasses the wavelet pipeline entirely. Consequently, the output includes all features of the
original embedding - including components that were projected out during decomposition. For this reason, the returned signal is comparable to a
wavelet-based reconstruction only after re-projecting its features into the same embedding subspace.
:::


```python
#| eval: false
percentile_emb_decomp.create_wreconstruction( # reconstruct a signal from specified frequency bands (fbands)
    subjs=[(40, 42), 50],                     # Subjects to project the reconstruction for: may be specified as intervals, e.g., [(40, 42)]
    fbands=[]                                 # frequency bands to include in the reconstruction (default is 'all')
).features_tint['subj40']
```

::: {.callout-caution}
The original (undecomposed) signal and the fully reconstructed signal are not perfectly identical. Small differences arise from boundary handling, finite-precision arithmetic, and other numerical effects in the wavelet transform.
:::


```python
#| eval: false

percentile_emb_decomp.create_wreconstruction( 
    subjs=[(40, 42), 50],              
    fbands='all'                    
).features_tint['subj40'] - percentile_emb_decomp.create_wreconstruction(
    subjs=[(40, 42), 50],                 
    fbands=[]                           
).features_tint['subj40'][[0,10,18]]  # We have to re-project the features into the embedding space of the reconstruction
```
