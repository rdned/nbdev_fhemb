# `Piece` API
> `Piece` is a top‑level class that serves as the entry point for working with a specific performance segment, bundling its metadata, subjects, and time interval into a single object

#### The Method Resolution Order (MRO) 
>the exact linear path Python follows when looking up attributes or when super() is used.

```raw
Piece
 └── DBGateway
      └── Metadata
           └── PerformanceLibMixin
                └── ValidateSubjsMixin
                     └── object
```

#### Architectural responsibilities

- **Scope and policy owner**: declare which recording, which time window, and which subjects are in scope; hold piece parameters (temperature threshold, percentile interval).

- **DB lazy attributes**: trigger and own DB‑loaded lazy attributes of type `Datasource` (e.g., `face_t`, `facebbox_t`) via the `DBGateway`; these raw artifacts are the primary inputs for downstream derivation.

- **Router**: route requests to `Datasource` and factories and ensure the final `Embedding` is produced under the `Piece`’s policies.

- **Guarantee**: ensure consistent `Piece`‑level parameters so results for the same `Piece` are comparable and reproducible.

## Lazy-loaded Piece *properties*

Available attributes (*readonly*) include the following **`Datasource` objects**:

- region of interest: `roi`, `roi_t`

- face bounding box: `facebbox`, `facebbox_t`

- face position: `faceposition_t`

:::{.callout-caution}
Pairs like `roi`/`roi_t` and `facebbox`/`facebbox_t` differ slightly because they originate from different tracking backends.
::: 

:::{.callout-note}
The lazy properties are loaded on first access through `DBGateway` and then cached.
:::
