# Embedding Videos
> Embedding Videos is a dedicated module-level example. It is useful primarily for debugging the [first step](./module_functions_ex.ipynb#obtain_input_data) of the basic workflow, as it displays the `Datasource` signal - effectively a video - in sync with any Embedding constructed from it. This synchronized view helps verify that the embedding reflects the underlying temporal structure of the data and makes it easier to diagnose issues in preprocessing, alignment, or feature extraction.

## Synchronize Audio with an `Embedding` Time Series
> for a chosen subject 40 and a feature 0th eigenface (principal component 0) 

```python
#| eval: false

makeVideo(
    interval=(5,25),                                    # Video interval [sec] 
    interval_shift = round(dg0.time_interval[0]/25),    # A shift to sync the audio and video [sec]
    feature=emb_prj_decomp_rec1.get_ts_subjs('pc0'),    # Feature time series
    subj=[40],                                          # Subject ID
    title=f"The 0th percentile in the frequency band {emb_prj_decomp_rec1.frequencies['pcomponents_ts']} Hz",
)
```

## Synchronizing a `Datasource` Signal with Its `Embedding` Time Series

### A chosen subject and an `Embedding` and two different `Datasource`s
> subject 40, 1st binheight, and region of interest and facial bounding box datasources.

```python
#| eval: false
emb_prj_decomp_rec1.binedges['binheights_ts']['subj40'][1]   # 'bh1' interval [st Celsius] 
```

```python
#| eval: false

makeVideo(
    dg0.roi.features_tint,                             # Region of interest for the animation   
    dg0.facebbox_t.features_tint,                      # Facial bounding box for the animation 
    interval=(5,25),                                   # Video interval [sec]
    interval_shift = round(dg0.time_interval[0]/25),   # A shift to sync the audio and video [sec]  
    feature=emb_prj_decomp_rec1.get_ts_subjs('bh1'),   # Feature time series
    subj=[40],                                         # Subject ID
    title=f"The 1th binheight in the frequency band {emb_prj_decomp_rec1.frequencies['binheights_ts']} Hz"
)
```

### Two chosen subjects and an `Embedding` and a single `Datasource`
> subjects 40 and 41, 0th eigenface, and region of interest datasource

```python
#| eval: false

makeVideo(
    dg0.roi_t.features_tint,                           # Region of interest for the animation
    interval=(5,25),                                   # Video interval [sec]
    interval_shift = round(dg0.time_interval[0]/25),   # A shift to sync the audio and video [sec]
    feature=emb_prj_decomp_rec1.get_ts_subjs('pc0'),   # Feature time series
    subj=[40, 41],                                     # Subject ID
    title=f"0th eigenfaces in the frequency band {emb_prj_decomp_rec1.frequencies['pcomponents_ts']} Hz",
)
```

### Two chosen subjects and an `Embedding` and a two `Datasource`s
> subjects 40 and 41, 10th percentile, and region of interest and facial bounding box datasources

```python
#| eval: false

makeVideo(
    dg0.roi.features_tint,                            # Region of interest for the animation
    dg0.facebbox_t.features_tint,                     # Facial bounding box for the animation
    interval=(5,25),                                  # Video interval [sec]
    interval_shift = round(dg0.time_interval[0]/25),  # A shift to sync the audio and video [sec]
    feature=percentile_emb.get_ts_subjs('p10'),       # Feature time series
    subj=[40, 41],                                    # Subject IDs  
    title=f"10th percentiles in the frequency band {percentile_emb.frequencies} Hz",
)
```
