# Music and Emotions

This simplification keeps the text uncluttered while preserving the essential meaning, allowing us to focus on the analysis and interpretation rather than on repetitive terminology.


## Introduction

The goal of this project is to deepen the understanding of the relationship between music and emotions by analyzing audience reactions during a classical opera performance. We use a time series of facial heatmaps capturing audience reactions, alongside synchronized audio embeddings, to model and interpret emotional responses.


## Methodology

### Tracking

**Facial heatmaps**: Track individual heatmaps corresponding to individual faces in the audience during the opera.

**Storage**: Store these time series in a database (DBMS) specialized for time series data.

### Embedding

**Computation**: Calculate embeddings (stochastic processes) for each tracked heatmap.

**Objective**: Extract stochastic processes that minimize noise while preserving the underlying stochastic patterns that capture the audience’s emotional responses to the music.


### Audio Embedding

**Computation**: Calculate embeddings of the synchronized audio from the opera.

**Alignment**: Ensure temporal alignment between audio embeddings and facial heatmap embeddings.

### Connecting Music and Emotions

**Model**: Implement a classifier to connect the stochastic processes derived from audience emotions with the music embeddings.

**Training objective**: Train the classifier to predict emotional states from music features and previous emotional states.

**Causal hypothesis**: Use prediction accuracy and robustness as indirect evidence for a causal influence of music on emotions.

### Explainable AI Analysis

**XAI methods**: Apply explainable AI techniques to the trained ANN.

**Feature importance**: Identify crucial input features (music characteristics or previous emotional states) that drive emotion prediction.

**Insights**: Gain interpretable insights into how specific musical features influence emotional responses over time.





## Outputs obtained

### Tracking

Tracked individual facial heatmap time series for audience members are stored in a DBMS.

### Embedding software

**FhEmb** software package takes on input the heatmap time series from the DBMS and creates an embedding time series that is used for further analysis. 

### Extracted embeddings (yield the main findings, see the Sect.6 below)

**Patterns**: We aim at extracting embeddings that highlight underlying stochastic patterns capturing the audience’s emotional responses to the music.

**Noise reduction**: Such embeddings reduce noise while preserving emotionally relevant structure.


## Aimed results

**Linking music and emotion**: A robust model linking musical features to emotional reactions.

**Temporal dynamics**: A deeper understanding of the temporal dynamics of emotions during live music.

**Explainable relationships**: Explainable insights into potential causal relationships between music and emotions.



## Future work – possible extensions

**Generality**: Extend the approach to other music genres and live performance settings.

**Multimodal data**: Incorporate physiological data (e.g., heart rate, skin conductance) for multimodal emotion analysis.

## Main Findings

### Signals of emotional dynamics, related to opera performance.

This claim is supported by the following evidence:

#### 1. Synchronization classification of signals of individual faces
We’ve been able to build classifiers that achieve expected classification accuracy in distinguishing of synchronized vs non-synchronized signals. Specifically, classifiers

- can recognize with accuracy well above chance whether listeners were exposed to opera sections that are identical vs musically or dramatically very different;
- struggle to distinguish whether two listeners were exposed to sections that are identical vs musically and dramatically similar.

This suggests that such signals are sensitive to coarse-grained musical and dramatic differences, but less so to fine-grained ones.

#### 2. Clustering of signals of individual faces
We’ve been able to define clustering that is well aligned with opera structure. Specifically,

- signals of an opera segment are clustered together;
- signals within a single opera segment create very few sub-clusters: typically one main cluster and 1 or 2 additional tiny and temporarily unstable clusters.

This indicates that such signals reflect the musical/dramatic structure.

### Facial heatmap time series appear to be constrained in a highly specific way
This claim is supported by the following evidence:

#### Signals that are quasi-functionally connected in each face during opera performance.
We’ve found signals that for each face are strongly stochastically dependent.

- Functional dependence beyond artifacts.
However, this stochastic dependence is so strong that it effectively is functional. The robustness of this dependence makes it difficult to attribute it solely to algorithmic arti facts.

- It is not clear however whether this constraint is induced exclusively by exposition to an opera performance, or perhaps to any emotional stimulus (joy, empathy), or by a more generic physiological invariant (e.g., a marker of autonomic regulation or overall health).

These findings suggest that, alongside music-driven emotional dynamics, there may be additional, highly structured components in the facial heatmap signals whose origin is not yet fully understood and warrants further investigation.
