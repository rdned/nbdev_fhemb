# `EmbeddingCollection` API
> A container for multiple `Embedding` objects that provides dynamic attribute access, batch computations, and lagged‑correlation utilities across embeddings.

#### The Method Resolution Order (MRO) 
>the exact linear path Python follows when looking up attributes or when super() is used.

```raw
EmbeddingCollection
 └── object
```

#### Architectural responsibilities

- **Container of analytic units**: hold a typed set of `Embedding` objects (e.g., "raw", "pca", "wavelet", "reconstructed") and provide a unified interface for operating across them. The collection itself contains no feature×time matrices — it organizes embeddings that do.

- **Coordination layer**: act as the orchestration point for multi‑embedding workflows (e.g., computing lagged correlations, cross‑embedding comparisons, or synchronized projections). It ensures that operations spanning multiple embeddings remain consistent in time indexing, subject sets, and feature alignment.

- **Dynamic dispatch**: expose a single attribute/method namespace that forwards calls to the appropriate embeddings. When multiple embeddings implement the same method, the collection merges or wraps results into a coherent multi‑embedding output.

- **Cross‑embedding analytics**: provide higher‑level computations (lagged correlations, correlation graphs, heatmaps) that require simultaneous access to multiple embeddings. These operations are not responsibilities of individual embeddings; they belong to the collection as the multi‑embedding analytic layer.

- **Extensibility point**: allow incremental addition of embeddings (`.add`), enabling pipelines where new transforms (PCA, wavelets, reconstructions, filtered variants) are appended and immediately become available to downstream multi‑embedding analyses.

- **Crucial point**: whereas the system’s atomic analytic unit is the `Embedding`, the `EmbeddingCollection` is the relational analytic unit - it enables analyses that depend on comparing, correlating, or jointly inspecting multiple embeddings at once.
