# Create a `Wdecomposition` Object and an `EmbeddingCollection` from It

::: {.callout-tip}
`Wdecomposition` is not typically used directly by end users. Instead, it serves as an internal component that supports the reconstruction of wavelet‑derived subbands.
:::

## `Wdecomposition`: Decomposed Bin‑Height Embedding Across Frequency Bands

```python
#| eval: false
#| exports
from fhemb.utils.factories import wfactory
```

```python
#| eval: false
#| exports

wavelet_factory = wfactory(
    'db4',       # The type of wavelet to use: 'db4' is Daubechies wavelet transform of order 4. Other options include "db2", "sym2", "coif1", etc.
    8            # The number of decomposition levels
)
```

```python
#| eval: false
#| exports

from nbdev_fhemb.embedding_ex import bin_emb
```

```python
#| eval: false
#| exports

bin_emb_decomp = bin_emb.create_wdecomposition(  # decompose to frequency bands of the wavelet transform specified by 'factory'
    features=[('bh0', 'bh10')],                  # specify features of interest: binheights are prefixed by 'bh' and may be specified also as intervals, e.g., [('bh0, 'bh10)]
    factory=wavelet_factory                      # specify the wavelet transform for the decomposition
)
```

## Create an `EmbeddingCollection` Object from the `Wdecomposition` Object
> by passing the dictionary of embedding components to the `EmbeddingCollection` constructor 

```python
#| eval: false
#| exports
from fhemb.embedding import EmbeddingCollection
```

```python
#| exports
#| eval: false

bin_emb_decomp_coll = EmbeddingCollection(
    bin_emb_decomp.fband_embedding         # dictionary of embedding components  
)
```

::: {.callout-tip}
The resulting `EmbeddingCollection` object presents a specific embedding format that can be used for further analysis and visualization.
:::
