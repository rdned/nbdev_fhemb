# Embeddings
> are implemented in terms of a class `Embedding`, the atomic analytic unit, and a class `EmbeddingCollection`, the coordination layer that enables multi-embedding workflows.

Together, they form the foundation of the analytical architecture.

## `Embedding` class
> holds the concrete per‑subject feature×time matrices and minimal provenance needed for deterministic, composable analysis; every transformation in the pipeline - PCA, filtering, wavelets, reconstruction - both consumes and produces embeddings, making them the system’s universal interchange format. 

## `EmbeddingCollection` class
> is a generalization of the more elementary `Embedding` class. 

- It organizes multiple `Embedding` objects (of different types) and exposes a unified interface for cross‑embedding operations. 

- It enables analyses that depend on comparing, correlating, or jointly inspecting several embeddings at once, providing the orchestration and dynamic dispatch that individual `Embedding` objects intentionally avoid. 

- It can be created either 

  - from a set of existing embeddings via the `EmbeddingCollection` constructor, 
  
  - or directly from a single `Datasource` through `Datasource.project` method. 
  
  The first approach allows embeddings originating from different `Datasource` objects to be combined, while the second is less flexible but syntactically more concise.
